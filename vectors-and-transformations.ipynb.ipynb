## Importing Libraries:

```python
import numpy as np
import matplotlib.pyplot as plt
```

### Coeficients of matrix A:

```python
A = np.array([[2,3],[5,7]])
A
```

### Coefficents of Matrix B:

```python
B = np.array([[11,13],[17,19]])
B
```

>Solve for x:

```python
x = np.linalg.solve(A,B)
x
```

# Solve System of complex Equations:

#### Coefficients matrix A with complex no:

```python
A_complex = np.array([[1+2j,2-1j],[3-1j,4]])
A_complex
```

#### Coefficients matrix B with complex no:

```python
B_complex = np.array([4+1j,3+4j])
B_complex
```

```python
x_complex = np.linalg.solve(A_complex,B_complex)
x_complex
```

# Vectors:

#### Create a vector:

```python
v = np.array([2,3])
v
```

### Plot the vector:

```python
plt.figure()
plt.quiver(0,0,v[0],v[1],angles='xy',scale_units='xy',scale=1,color='r')
plt.xlim(0,5)
plt.ylim(0,5)
plt.grid()
plt.title('2D Vector')
plt.xlabel('x')
plt.ylabel('y')
plt.gca().set_aspect('equal',adjustable='box')
plt.show()
```

 **************************

# Linear Transformation:

>converting 45 degrees to radians:

```python
theta = np.radians(45)#define the rotation matrix for 45 degree rotation
rotation_matrix = np.array([[np.cos(theta),-np.sin(theta)],[np.sin(theta),np.cos(theta)]])
rotation_matrix
```

### Applying the transformation:

```python
v_transformed = np.dot(rotation_matrix,v)
```

```python
v_transformed
```

### Plotting the original and transformed vector:

```python
plt.figure()
plt.quiver(0,0,v[0],v[1],angles='xy',scale_units='xy',scale=1,color='r',label='original vector')
plt.quiver(0,0,v_transformed[0],v_transformed[1],angles='xy',scale_units='xy',scale=1,color='b',label='transformed vector')
plt.xlim(-1,5)
plt.ylim(-1,5)
plt.grid()
plt.title('2D Vector Transformation')
plt.xlabel('x')
plt.ylabel('y')
plt.gca().set_aspect('equal',adjustable='box')
plt.show()
```

# Shear Transformation:

```python
shear_factor = 0.5
shear_matrix = np.array([[1,shear_factor],[0,1]])
shear_matrix
```

#### Applying the Shear Transformation:

```python
v_Sheared = np.dot(shear_matrix,v)
v_Sheared
```

#### Plotting the original,rotated and sheared vectors:

```python
plt.figure()
plt.quiver(0,0,v[0],v[1],angles='xy',scale_units='xy',scale=1,color='r',label='Original Vector')
plt.quiver(0,0,v_transformed[0],v_transformed[1],angles='xy',scale_units='xy',scale=1,color='b',label='Rotated Vector')
plt.quiver(0,0,v_Sheared[0],v_Sheared[1],angles='xy',scale_units='xy',scale=1,color='g',label='Sheared Vector')
plt.xlim(-1,5)
plt.ylim(-1,5)
plt.grid(True)
plt.axhline(0,color='k',linewidth=0.5)
plt.axvline(0,color='k',linewidth=0.5)
plt.title('2D Vector Transformations')
plt.xlabel('x')
plt.ylabel('y')
plt.legend()
plt.gca().set_aspect('equal',adjustable='box')
plt.show()
```
